```
import pandas as pd
dataset = pd.read_csv('houseprices_project.csv')
```

```
from sklearn.preprocessing import OneHotEncoder

# Initialize the one hot encoding model
onehot_encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)

# Transform the data using the model
onehot_encoded = onehot_encoder.fit_transform(dataset[['ocean_proximity']])


# Print new feature names
new_features = onehot_encoder.get_feature_names_out(['ocean_proximity'])
print(new_features)
print(onehot_encoded)
#dsaflnnl;sdfjk
```

```
# Lets add these new columns to the dataset and drop the original column
# Drop original column
dataset.drop(['ocean_proximity'], axis=1)
dataframe = pd.DataFrame(onehot_encoded, columns=new_features)
dataset = pd.concat([dataset, dataframe], axis=1)
print(dataset)
```

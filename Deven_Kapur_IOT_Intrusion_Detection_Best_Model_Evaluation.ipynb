**LOAD THE MODEL**




```python
import pickle
#Set the path for Best Model out of all tested models
file_name=open('/content/drive/Shareddrives/ResearchPrograms_V2/Hasini_Vehical_Audio_Project/Emergency_Vehical_Siren_Data/Results/five_categories_best_model_random_forest','rb')
model=pickle.load(file_name)
print('Model is loaded')
```

**READ THE TEST DATASET**

```python
import pandas as pd

LABEL_COLUMN='LABEL'

#load the path of arjan's test dataset
TEST_PATH='/content/drive/Shareddrives/ResearchPrograms_V2/Hasini_Vehical_Audio_Project/Emergency_Vehical_Siren_Data/CSV/five_categories_audio_test.csv'
data_test=pd.read_csv(TEST_PATH)
data_test.head()
```

**LABEL ENCODING**

```python
# define the label column
LABEL_COLUMN = "LABEL"
```

```python
# get the labels
LABELS = list(data_test[LABEL_COLUMN].unique())
LABELS.sort()
print(LABELS)
```

```python
# encode the labels
data_test[LABEL_COLUMN] = pd.factorize(data_test[LABEL_COLUMN], sort = True)[0]
data_test[LABEL_COLUMN].value_counts()
```

**SEPERATE FEATURES AND LABELS**

```python
# all the rows
# all the columns without the last column
x_test = data_test.drop(LABEL_COLUMN, axis = 1).values
# all rows
# only the lastb column, which is the label
y_test = data_test[LABEL_COLUMN].values
```

**Use the best model to make predictions**

```python
# make predictions
predictions = model.predict(x_test)
print(predictions)
```

**EVALUATE THE MODEL PERFORMANCE ON TEST DATASET**

```python
#evaluation function

def model_evaluations(y_true, y_pred, labels):
  import matplotlib.pyplot as plt
  import seaborn as sns
  from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

  acc_score = accuracy_score(y_true, y_pred)
  print("Accuracy score: {}\n".format(acc_score))

  print("Classification Report: {}".format(classification_report(y_true, y_pred)))

  plt.figure(figsize = (10,10))
  sns.heatmap(confusion_matrix(y_true, y_pred),  annot = True, fmt="g", cmap = "Blues", xticklabels = labels, yticklabels = labels)
  plt.title("Consfuion Matrix")
  plt.show()
```

```python
# run the evaluation functions
model_evaluations(y_test, predictions, LABELS)
```
